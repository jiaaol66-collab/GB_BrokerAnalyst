```python
#3.清洗日期（部分需要到excel内处理,CIY）
def parse_dates(date_series,formats):
    date_formats = ['%Y-%m-%d', '%Y.%m.%d', '%Y%m%d', '%Y/%m/%d']
    #formats是列表形式，可自定义添加存在的时间格式
    date_formats += formats
    for fmt in date_formats:
        # 尝试解析日期
        parsed_dates = pd.to_datetime(date_series, format=fmt, errors='coerce')
        # 合并成功解析的日期
        date_series = date_series.mask(parsed_dates.notna(), parsed_dates)
    return pd.to_datetime(date_series, errors='coerce')

extra = pd.read_excel(r'2.')
df_1 = pd.concat([df1,extra],ignore_index=True)

formats = ['%Y 年 %m 月 %d 日','%Y年%m月%d日','%Y.%m','%Y年%m月%d号','%Y月%m月%d日']#DIY
df1['推荐日期清洗版'] = parse_dates(df1['推荐日期'],formats)

print(df1[df1['推荐日期清洗版'].isnull()]['推荐日期'].value_counts())#这些稍后在excel内优先处理

```
