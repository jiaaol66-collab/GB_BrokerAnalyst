```python
import os 
import pandas as pd
```

```python
output_dir = '头部公募基金整合'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
```

```python
def traverse_folder_headfund(root_folder):
    # 遍历根文件夹下的所有子文件夹和文件
    for subdir, dirs, files in os.walk(root_folder):
        for file in files:
            # 选取所有xls(x)文件
            if '.xls' in file.lower():
                file_path = os.path.join(subdir, file)
                print(f"正在提取: {file_path}")
                try:
                    xls = pd.ExcelFile(file_path)
                    df = pd.read_excel(xls, sheet_name='sheet1')
                    df['文件名'] = file.rstrip('.xlsx')
                    df_concat.append(df)
                except Exception as e:
                    print(f"无法读取文件 {file_path}: {e}")

df_concat = []
folder_path = '头部公募基金'
traverse_folder_headfund(folder_path)
df0 = pd.concat(df_concat)

print('success')
                        
```

```python
file_path = os.path.join('头部公募基金整合','头部公募基金数据.xlsx')
df0.to_excel(file_path,index=False)
import os
import platform
def open_file(file_path):
    if platform.system() == "Windows":
        os.startfile(file_path)
    elif platform.system() == "Darwin":
        os.system(f'open "{file_path}"')
    elif platform.system() == "Linux":
        os.system(f'xdg-open "{file_path}"')
    else:
        print(f"无法在 {platform.system()} 系统上打开文件。请手动查找并打开 {file_path}。")

open_file(file_path)
open_file('公募基金复制列.xlsx')
```
