```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import time, os, sys, shutil
from datetime import datetime, timedelta
from matplotlib.dates import date2num


start = time.time()


def make_dir(path):
    if not os.path.exists(path):
        os.makedirs(path)


def copy_file(src, dst, tag_=0):       # 从源路径（src)复制一个文件到目标路径(path)，
    tag_update = 0
    if not os.path.exists(src):
        print(f'源文件{src}不存在！无法复制。')
    else:
        if tag_ == 1:  # 当tag_ = 1时，如果文件已存在则进行替换(多为输出的文件)，其中代码和说明文件不进行替换。(除了更新文件)
            if os.path.exists(dst):
                tag_update = 1
                os.remove(dst)
        if not os.path.exists(dst):
            shutil.copyfile(src, dst)
            if tag_ == 1:
                if tag_update == 1:
                    print(f'已更新{src}--->{dst}')
                else:
                    print(f'已复制{src}--->{dst}')
            else:
                print(f'已复制{src}--->{dst}')
        else:
            print(f'{dst}已有文件，不进行复制')


plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
with open('../参评信息/年份信息.txt', 'r') as f:
    lines = f.readlines()

for line in lines:
    yr = int(line)
    break

print('评奖年份:', yr)
```

```python
if yr == 2024:
    score = {'年化收益': 0.3, '风险调整后收益': 0.25, '超大盘收益率': 0.15, '最大回撤': 0.15, '机构占比': 0.075,
                 '券商自持': 0.075}
    
elif yr == 2025:
    score = {'年化收益': 0.3, '风险调整后收益': 0.25, '超大盘收益率': 0.15, '最大回撤': 0.2, '机构占比': 0.1}
```

```python
i = 1
for prod_cate, prod_nv in nv_all.groupby(['投资类型','投资策略','产品全称','时间跨度']):
    tspan = prod_cate[3]
    if tspan == 5:
        term = '5年期'
        prod_nv = prod_nv[prod_nv['净值日期'].between(f'{yr-5}-1-1',f'{yr-1}-12-31')]
        tag = 1
    elif tspan == 3:
        term = '3年期'
        prod_nv = prod_nv[prod_nv['净值日期'].between(f'{yr-3}-1-1',f'{yr-1}-12-31')]
        tag = 1
    elif tspan == 1:
        term = '1年期'
        prod_nv = prod_nv[prod_nv['净值日期'].between(f'{yr-1}-1-1',f'{yr-1}-12-31')]
        tag = 1
    elif tspan == 0:
        term = '未入选'
        tag = 0
        
    output_dir = f'../输出文件/可视化/{prod_cate[0]}/{prod_cate[1]}/{term}'
    make_dir(output_dir)
    print(i,'/',len(nv_all[['产品全称','时间跨度']].drop_duplicates()))
    i += 1
    company_name = prod_nv['公司名称'].unique()[0]
    
    fig, ax = plt.subplots(figsize=(22, 10))
    
    line, = ax.plot(prod_nv['净值日期'], prod_nv['累计净值'],
                                marker='o', markersize=1,
                                linewidth=1.5, alpha=0.8,
                                label=f"{prod_cate[2]}",
                                zorder=2)
    
    
    if tag == 1:
        rank = int(prod_nv['总分排名'].unique()[0])
        R = f'排名{rank}_'
        abnormal = int(prod_nv['异常次数'].unique()[0])
        freq = prod_nv['净值频率'].unique()[0]
        org_metrics = {}
        for label in score.keys():
            org_metrics[label] = prod_nv[label].unique()[0]
        org_metrics['排名'] = rank
        org_metrics['异常次数'] = abnormal
        org_metrics['年期'] = term
        org_metrics['频率'] = freq

        # === 右侧指标展示板 ===
        if yr == 2024:
            metric_text = (
                f"产品综合指标\n"
                f"------------------------\n"
                f"排名: {org_metrics['排名']}\n"
                f"年化收益: {org_metrics['年化收益']*100:.2f}%\n"
                f"风险调整后收益: {org_metrics['风险调整后收益']:.2f}\n"
                f"超大盘收益率: {org_metrics['超大盘收益率']*100:.2f}%\n"
                f"最大回撤: {org_metrics['最大回撤']*100:.2f}%\n"
                f"机构占比: {org_metrics['机构占比']*100:.2f}%\n"
                f"券商自持: {org_metrics['券商自持']*100:.2f}%\n"
                f"异常次数: {org_metrics['异常次数']}\n"
                f"频率: {org_metrics['频率']}\n"
                f"年期: {org_metrics['年期']}\n"
                f"投资类型:{prod_cate[0]},投资策略:{prod_cate[1]}"
            )
        if yr == 2025:
            metric_text = (
                f"产品综合指标\n"
                f"------------------------\n"
                f"排名: {org_metrics['排名']}\n"
                f"年化收益: {org_metrics['年化收益']*100:.2f}%\n"
                f"风险调整后收益: {org_metrics['风险调整后收益']:.2f}\n"
                f"超大盘收益率: {org_metrics['超大盘收益率']*100:.2f}%\n"
                f"最大回撤: {org_metrics['最大回撤']*100:.2f}%\n"
                f"机构占比: {org_metrics['机构占比']*100:.2f}%\n"
                f"异常次数: {org_metrics['异常次数']}\n"
                f"频率: {org_metrics['频率']}\n"
                f"年期: {org_metrics['年期']}\n"
                f"投资类型:{prod_cate[0]},投资策略:{prod_cate[1]}"
            )
        # 在右侧创建文字框
        plt.figtext(0.86, 0.65, metric_text,
                    ha='left', va='top',
                    fontsize=16,
                    bbox=dict(boxstyle='round',
                              facecolor='white',
                              alpha=0.8,
                              edgecolor='grey'))
    else:
        R = ''
        
    # 调整图例样式和布局
    handles, labels = ax.get_legend_handles_labels()

    # 图例数量控制逻辑
    max_legend_items = 10
    if len(handles) > max_legend_items:
        handles = handles[:max_legend_items]
        labels = labels[:max_legend_items]
        if labels[0] == '拟合净值':
            labels[-1] = f"{labels[-1]} (剩余{prod_count-max_legend_items+1}不显示)"  # 添加数量提示
        else:
            labels[-1] = f"{labels[-1]} (剩余{prod_count-max_legend_items}不显示)"

    if handles:
        ax.legend(handles, labels,
                  loc='upper left',
                  bbox_to_anchor=(1.02, 1),
                  borderaxespad=0.,
                  frameon=False,
                  fontsize=10,
                  handlelength=1.5,
                  labelspacing=0.8)

    plt.xticks(rotation=45)
    plt.ylabel("累计净值", fontsize=12)
    plt.grid(True, alpha=0.3)
         
    # 坐标轴格式化
    ax.xaxis.set_major_locator(mdates.MonthLocator(bymonth=(1, 7)))
    ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m'))
    # ========== 添加考察期截止线 ==========
    ax.axvline(x=datetime(yr - 1, 12, 31), color='darkred', linestyle='--', linewidth=2, alpha=0.7)
    ax.annotate('',
                xy=(datetime(yr - 1, 12, 31), ax.get_ylim()[1] * 0.95),
                xycoords='data',
                color='darkred'
                )
    ax.axvline(x=datetime(yr - 2, 12, 31), color='blue', linestyle='--', linewidth=2, alpha=0.9)
    ax.annotate('',
                xy=(datetime(yr - 2, 12, 31), ax.get_ylim()[1] * 0.95),
                xycoords='data',
                color='blue'
                )

    if term in ['5年期','3年期']:
        ax.axvline(x=datetime(yr - 4, 12, 31), color='green', linestyle='--', linewidth=2, alpha=0.7)
        ax.annotate('',
                    xy=(datetime(yr - 4, 12, 31), ax.get_ylim()[1] * 0.95),
                    xycoords='data',
                    color='green'
                    )
    if term in ['5年期']:
        ax.axvline(x=datetime(yr - 6, 12, 31), color='orange', linestyle='--', linewidth=2, alpha=0.9)
        ax.annotate('',
                    xy=(datetime(yr - 6, 12, 31), ax.get_ylim()[1] * 0.95),
                    xycoords='data',
                    color='orange'
                    )
        
    fig.set_size_inches(22, 10)  # 加宽画布
    plt.subplots_adjust(right=0.8)  # 右侧留出20%空间
    plt.title(f"{prod_cate[2]}产品净值走势", pad=20)
    plt.tight_layout()
    plt.savefig(f'{output_dir}/{R}{company_name}_{prod_cate[2]}_{prod_cate[0]}_{prod_cate[1]}.png',
                bbox_inches='tight', dpi=80)
    plt.close()

    
    
    
```
